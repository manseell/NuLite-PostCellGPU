```python
import json
from shapely.geometry import shape

with open("/home/jovyan/path-vol/NuLite-PostCellGPU/Output/20-5312 PDL-1 22C3 Ab/cell_detection/cell_detection.geojson", "r") as f:
    data = json.load(f)

totale = 0
for elemento in data:
    geom = shape(elemento["geometry"])
    n = len(geom.geoms)
    classe = elemento["properties"]["classification"]["name"]
    print(f"{classe}: {n} nuclei")
    totale += n

print(f"\nTotale nuclei: {totale}")
```

```python
import os
import json
from shapely.geometry import shape


for directoryname in os.listdir("./Output/Output/"):
    dir_path = os.path.join("./Output/Output/" + directoryname)

    if not os.path.isdir(dir_path):
        continue

    cell_det_dir = os.path.join(dir_path, "cell_detection")

    if not os.path.isdir(cell_det_dir):
        continue

    totale = 0

    for filename in os.listdir(cell_det_dir):
        if filename.endswith("cell_detection.geojson"):
            file_path = os.path.join(cell_det_dir, filename)

            with open(file_path, "r") as f:
                data = json.load(f)

            for elemento in data:
                geom = shape(elemento["geometry"])
                n = len(geom.geoms)
                classe = elemento["properties"]["classification"]["name"]
                print(f"{classe}: {n} nuclei")
                totale += n

    print(f"Totale nuclei in {directoryname}: {totale}")
    print("")
```
